## 9. Part B — Add the Flex Sensor and MCP3008

In Part A, the Raspberry Pi controlled an LED using a **digital GPIO output**.

In this part, we add a **flex sensor** so that the Raspberry Pi can react to how much the sensor is bent.

The flex sensor behaves like a variable resistor: its resistance changes when it bends. However, the Raspberry Pi GPIO pins cannot directly measure this continuously changing analog value. For this reason, we use an **MCP3008 Analog-to-Digital Converter (ADC)**.

The signal path is:

**Flex sensor → voltage divider → MCP3008 → Raspberry Pi → LED**

The flex sensor and a fixed resistor first create a voltage that changes with bending. The MCP3008 measures this voltage and sends the reading to the Raspberry Pi through SPI.

---

### 9.1 MCP3008 pinout

The MCP3008 has 16 pins. In this tutorial, only one analog input channel, **CH0**, is required.

When the notch of the MCP3008 is facing upward, pin 1 is at the upper-left side.

### 9.2 A note about the flex sensor

The flex sensor is delicate near its terminal/base area.

Keep the base as straight and stable as possible and bend only the flexible sensing section. Avoid sharply folding or twisting the sensor.

The two terminals of the flex sensor do not have a positive or negative polarity for this experiment.

---

### 9.3 How the flex sensor is connected

The flex sensor and a fixed resistor form a **voltage divider**.

The basic connection is:

**3.3 V → Flex Sensor → Measurement Point → Fixed Resistor → GND**

The **measurement point** is also connected to **CH0 (pin 1)** of the MCP3008.

Therefore, three connections meet at the measurement point:

- one terminal of the flex sensor,
- one side of the fixed resistor,
- MCP3008 CH0.

When the flex sensor bends, its resistance changes. This changes the voltage at the measurement point, which is then measured by the MCP3008.

---

### 9.4 Example breadboard connection used in this tutorial

The following example shows the breadboard arrangement used and tested for this tutorial.

#### MCP3008 placement

Place the MCP3008 across the centre gap of the breadboard, starting at row 27.

| MCP3008 pin | Breadboard position |
|---:|---|
| 1 — CH0 | F27 |
| 2 | F28 |
| 3 | F29 |

| 12 — DOUT | E31 |
| 13 — CLK | E30 |
| 14 — AGND | E29 |
| 15 — VREF | E28 |
| 16 — VDD | E27 |

#### MCP3008 to Raspberry Pi

Connect:

- **E27 / pin 16 (VDD)** → Raspberry Pi **3.3 V, physical pin 1**
- **E28 / pin 15 (VREF)** → Raspberry Pi **3.3 V, physical pin 17**
- **E29 / pin 14 (AGND)** → Raspberry Pi **GND, physical pin 9**
- **E30 / pin 13 (CLK)** → Raspberry Pi **GPIO11 / physical pin 23**
- **E31 / pin 12 (DOUT)** → Raspberry Pi **GPIO9 / physical pin 21**
- **E32 / pin 11 (DIN)** → Raspberry Pi **GPIO10 / physical pin 19**
- **E33 / pin 10 (CS/SHDN)** → Raspberry Pi **GPIO8 / physical pin 24**
- **E34 / pin 9 (DGND)** → Raspberry Pi **GND, physical pin 14**

#### Flex sensor and resistor

The 3.3 V connection from row 27 is shared with the flex-sensor circuit:

- Connect **A27 → A40** with a jumper wire.
- Place one flex-sensor terminal at **E40**.
- Place the other flex-sensor terminal at **E41**.
- Place one leg of the fixed resistor at **B41**.
- Place the other resistor leg at **B49**.
- Connect **A49 → Raspberry Pi GND, physical pin 39**.
- Connect **A41 → F27**, which is MCP3008 **CH0**.

Because holes A–E in the same numbered breadboard row are electrically connected:

- row 40 carries **3.3 V**,
- row 41 becomes the **measurement point**,
- row 49 is connected to **GND**.

The resulting path is:

**3.3 V → Flex Sensor → Row 41 → Fixed Resistor → GND**

and:

**Row 41 → MCP3008 CH0**

## 10. Check the Flex-Sensor Reading

The next cell reads MCP3008 channel 0.

Keep the sensor straight first and observe the value. Then bend only the flexible sensing area and observe how the value changes.

The normalized reading is between **0.0 and 1.0**.

````python
from gpiozero import MCP3008
from time import sleep

flex = MCP3008(channel=0)

print("Flex-sensor readings:")
for _ in range(20):
    print(f"{flex.value:.3f}")
    sleep(0.25)

flex.close()
````

## 11. First Flex-Sensor Test — Bend = LED ON

This first test keeps the behaviour intentionally simple.

1. Keep the flex sensor straight when the cell starts.
2. The notebook records a short baseline.
3. Bend the sensing area.
4. When the reading changes sufficiently from the baseline, the LED turns **ON**.
5. When the sensor returns close to its initial position, the LED turns **OFF**.

This is a basic functional test before introducing gradual brightness.


````python
from gpiozero import MCP3008, LED
from time import sleep

flex = MCP3008(channel=0)
led = LED(17)

print("Keep the flex sensor straight...")
sleep(2)

samples = [flex.value for _ in range(20)]
baseline = sum(samples) / len(samples)

print(f"Baseline value: {baseline:.3f}")
print("Now bend and release the sensor. The test will run for 15 seconds.")

threshold = 0.05

for _ in range(150):
    reading = flex.value
    change = abs(reading - baseline)

    if change > threshold:
        led.on()
        state = "BENT → LED ON"
    else:
        led.off()
        state = "STRAIGHT → LED OFF"

    print(f"Reading: {reading:.3f} | Change: {change:.3f} | {state}", end="\r")
    sleep(0.1)

led.off()
led.close()
flex.close()
print("\nFlex-sensor ON/OFF test completed.")
````

## 12. From Binary Control to Gradual Control

The previous test treated the sensor in a simple binary way:

- approximately straight → LED OFF
- sufficiently bent → LED ON

Human descriptions are often more gradual. A bend might be described as:

- almost straight
- slightly bent
- moderately bent
- strongly bent

These descriptions are examples of **linguistic terms**.

Instead of forcing every sensor reading into only ON or OFF, the next experiment uses the amount of bending as a gradual value between 0 and 1 and maps that value to LED brightness.


## 13. Calibrate Straight and Bent Positions

Before controlling brightness, the notebook needs two reference values:

- the reading when the sensor is **straight**
- the reading when the sensor is **strongly but safely bent**

The code works whether the ADC value increases or decreases during bending.


````python
from gpiozero import MCP3008
from time import sleep

flex = MCP3008(channel=0)

def average_reading(sensor, samples=30, delay=0.03):
    values = []
    for _ in range(samples):
        values.append(sensor.value)
        sleep(delay)
    return sum(values) / len(values)

print("Keep the flex sensor straight.")
sleep(3)
straight_value = average_reading(flex)
print(f"Straight reference: {straight_value:.3f}")

print("\nNow bend the flexible sensing area strongly but safely.")
sleep(3)
bent_value = average_reading(flex)
print(f"Bent reference: {bent_value:.3f}")

if abs(bent_value - straight_value) < 0.02:
    print("\nThe two reference values are very close. Check the wiring and repeat the calibration.")
else:
    print("\nCalibration completed.")

flex.close()
````

## 14. Fuzzy-Inspired Gradual LED Brightness

The next cell maps the calibrated bend amount continuously:

- **0.0** → approximately straight → LED off/very dim
- **0.5** → intermediate bend → medium brightness
- **1.0** → strongly bent → maximum brightness

It also calculates three overlapping linguistic memberships:

- **Straight**
- **Medium Bend**
- **Strong Bend**

A single bend value can partially belong to more than one linguistic term at the same time. This gradual overlap introduces the main intuition behind fuzzy linguistic variables.


````python
from gpiozero import MCP3008, PWMLED
from time import sleep

flex = MCP3008(channel=0)
led = PWMLED(17)

# Use the straight_value and bent_value obtained in the calibration cell.
difference = bent_value - straight_value

if abs(difference) < 0.02:
    raise ValueError("Calibration range is too small. Run the calibration cell again.")

def clamp(value, low=0.0, high=1.0):
    return max(low, min(high, value))

def bend_amount(reading):
    # Works whether the ADC reading rises or falls when the sensor bends.
    return clamp((reading - straight_value) / difference)

def memberships(x):
    straight = clamp(1 - 2*x)
    medium = clamp(1 - 2*abs(x - 0.5))
    strong = clamp(2*x - 1)
    return straight, medium, strong

print("Bend and release the flex sensor. The test will run for 20 seconds.")

for _ in range(200):
    reading = flex.value
    bend = bend_amount(reading)

    straight_m, medium_m, strong_m = memberships(bend)

    # Continuous physical output
    led.value = bend

    print(
        f"ADC: {reading:.3f} | Bend: {bend:.2f} | "
        f"Straight: {straight_m:.2f} | Medium: {medium_m:.2f} | "
        f"Strong: {strong_m:.2f} | LED brightness: {led.value:.2f}",
        end="\r"
    )

    sleep(0.1)

led.off()
led.close()
flex.close()
print("\nGradual flex-sensor experiment completed.")
````

## 15. What This Demonstrates

You have now moved through three levels of physical interaction:

1. **LED output:** the Raspberry Pi controls a physical device.
2. **Flex-sensor input:** the Raspberry Pi receives information about a physical action.
3. **Input-to-output mapping:** the amount of bending controls LED brightness.

The final experiment also introduces a key Computing with Words idea: a physical measurement can be interpreted through overlapping linguistic descriptions rather than only rigid ON/OFF categories.

This is an introductory demonstration. More complete fuzzy systems can later use membership functions, rule sets, inference, aggregation, and defuzzification.


## 16. Troubleshooting

### LED does not turn on
- check the LED polarity
- check GPIO17 and GND
- confirm that the LED has a 220–330 Ω series resistor

### Flex-sensor value does not change
- confirm that SPI is enabled
- check the MCP3008 orientation
- check 3.3 V and GND
- confirm that the voltage-divider midpoint is connected to MCP3008 CH0
- make sure the flex sensor is being bent along its sensing area, not at the terminals

### MCP3008 cannot be accessed
- verify the SPI wiring
- verify that SPI was enabled through `raspi-config`
- reboot the Raspberry Pi after enabling SPI if required

### LED brightness appears reversed
The calibration code is designed to handle increasing or decreasing ADC readings. Re-run the straight/bent calibration and make sure the straight and bent positions were held correctly.

### Readings fluctuate slightly
Small changes are normal for analog measurements. The calibration cells average several samples to make the experiment more stable.
