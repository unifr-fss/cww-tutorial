# Tutorial 1 - Soft Circuits Introduction

**Beginner’s Guide to E-Textiles / Soft Circuits**

This guide explores the different types of e-textile components available and demonstrates how to sew your own soft circuits using conductive thread. The demonstrations featured involve basic circuits without a microcontroller — one can be introduced later, but firstly understanding the basics of designing and making a circuit is essential.


## Learning goals

By the end of this tutorial, you should be able to:

- describe what soft circuits, e-textiles, and wearable electronics are;
- identify beginner-friendly sewable components and materials;
- plan a simple circuit on fabric before sewing;
- sew a basic LED circuit using conductive thread;
- explain how a parallel LED circuit differs from a single LED circuit;
- add a push button or switch to make the circuit interactive;
- test and troubleshoot a soft circuit safely.

## What are soft circuits, e-textiles, and wearable electronics?

A **circuit** is a closed path that allows electricity to flow. In a basic circuit, electricity travels from a power source, through connected components, and back to the power source.

Most circuits are made with wires. In **soft circuits**, the same idea is created with flexible conductive materials such as conductive thread or conductive fabric. This allows circuits to be sewn or integrated into textiles.

**E-textiles** are fabrics that include electronic components or electronic behaviour. **Wearable electronics** are electronic systems integrated into clothing, accessories, jewellery, costumes, or other wearable objects.

Soft circuits can be used for visual effects, interactive textiles, movement-based designs, body-related sensing, and other wearable technology projects.

## What are sewable components?

Electronic components are the parts that make a circuit work. In soft circuits, many components are designed with **sewable pins** or conductive pads. Conductive thread is wrapped or stitched around these pins so current can pass from one component to another.

A component does not always have to be specially labelled as “sewable” to be used in an e-textile project. However, it must be compatible with the voltage and the rest of the circuit.

**Key idea:** the thread must make strong contact with the metal/conductive part of each component.

## Choosing the right fabric

The fabric must be stable enough to support the components, but not so thick that it becomes difficult to sew through.

For beginners, **felt** is a good starting material because it is easy to cut, does not fray easily, and is simple to sew through.

Other possible fabrics include twill, lightweight denim, calico, drill fabric, or synthetic leather. If the fabric is too thin or flexible, it can be strengthened with iron-on fusing. For larger pieces of fabric, an embroidery hoop can help keep the surface stable while sewing.

## Conductive thread

Conductive thread looks similar to normal sewing thread, but it is coated with conductive material so it can carry electrical current.

When sewing, the thread should be wound several times around each sewable pin or conductive pad. This creates a stronger electrical connection.

> **Important:** conductive thread is not insulated like normal electrical wire. If two conductive thread paths touch or cross, the circuit may short or stop working. Keep positive (+) and negative (−) paths separate.

## Sewing equipment and additional materials

- **Craft scissors:** useful because conductive thread can be harder to cut than normal thread.
- **Wide-eye hand sewing needle:** easier to thread with thicker conductive thread.
- **Needle threader:** helpful if the conductive thread is thick or difficult to pass through the needle.
- **Chalk, wax, or fabric marker:** used to draw the circuit path before sewing.
- **Standard sewing thread:** secures unused component pins because it is non-conductive.
- **Masking tape or hot glue:** temporarily holds components in place while sewing.
- **Alligator clips:** test the circuit before sewing it permanently.
- **Clear nail polish or adhesive:** secures knots on the back side of the fabric.

## Beginner components

### Power supply

A soft circuit needs a power supply, usually a small coin cell battery and a battery holder. Many beginner sewable components are designed for low voltage, often around 3V.

Switch the power off or remove the battery while sewing or adjusting the circuit.

### LED

An LED is a light-emitting diode. It lights up when current flows through it in the correct direction.

Sewable LEDs usually have clear **positive (+)** and **negative (−)** markings. Standard LEDs can also be used, but their legs must be identified and bent into sewable loops.

### Push button or switch

A push button or switch controls when the circuit is complete. When the button is pressed or the switch is closed, current can flow and the LED can turn on.

## Making a standard LED sewable

Some projects use standard LEDs instead of sewable LED boards. To prepare a standard LED:

1. Identify the legs:
   - long leg = positive (+)
   - short leg = negative (−)
2. Mark one leg with a pen so you do not lose track.
3. Use needle-nose pliers to flatten the legs in opposite directions.
4. Twist each leg into a loop large enough for a needle to pass through.
5. Place the LED flat against the fabric before sewing.

Check that the LED voltage is compatible with the power supply.

## Before starting: plan and test

Before sewing, plan the circuit on paper or directly on the fabric.

1. Place the components where you want them.
2. Trace around each component.
3. Draw the positive (+) and negative (−) paths.
4. Check that the positive path connects only positive pins.
5. Check that the negative path connects only negative pins.
6. Make sure the two paths do not cross or touch.
7. Test the layout with alligator clips before sewing.

Testing first saves time and helps find mistakes before the components are sewn down.

# Demonstration 1: Simple LED Circuit

## Objective

Create a soft circuit that lights one LED using a battery holder and conductive thread.

## Circuit idea

The circuit is complete when:

```text
Battery positive (+) → LED positive (+)
LED negative (−) → Battery negative (−)
```

The LED will not light if the positive and negative sides are reversed.

<div style="display: flex; justify-content: center; margin: 16px 0;">
  <img src="Soft_computing_pictures/Soft10.png" width="850">
</div>

<div style="text-align: center; margin-bottom: 18px;">
  <b>Figure 10:</b> Simple LED Circuit Demonstration 
</div>

## Steps

1. Place the battery holder and LED on the fabric.
2. Make sure the positive (+) and negative (−) pins are easy to identify.
3. Mark the two sewing paths:
   - one path from battery (+) to LED (+)
   - one path from LED (−) back to battery (−)
4. Secure the components temporarily with tape or a small glue dot if needed.
5. Thread the needle with conductive thread and tie a neat knot.
6. Start on the back side of the fabric and sew through the battery’s positive (+) pin.
7. Loop around the positive pin 3–4 times so the thread firmly touches the metal.
8. Sew a running stitch toward the LED’s positive (+) pin.
9. Loop around the LED’s positive pin 3–4 times.
10. Knot off neatly on the back side and trim loose ends.
11. Use a new piece of conductive thread for the negative path.
12. Sew from the LED’s negative (−) pin to the battery’s negative (−) pin.
13. Knot off neatly and secure knots with clear nail polish or adhesive.
14. Insert or switch on the battery and check whether the LED lights.

## Checkpoint

The LED should light when the circuit is complete. If it does not, check polarity, loose knots, weak contact, or touching thread paths.

# Demonstration 2: Parallel LED Circuit

## Objective

Create a circuit with multiple LEDs connected to the same power source.

## Circuit idea

A **parallel circuit** gives each LED its own path across the same power source. Each LED receives the same voltage.

This is useful when multiple LEDs should light at the same time.

<div style="display: flex; justify-content: center; margin: 16px 0;">
  <img src="Soft_computing_pictures/Soft11.png" width="450">
</div>

<div style="text-align: center; margin-bottom: 18px;">
  <b>Figure 11:</b> Parallel LED Circuit Demonstration 
</div>

## Steps

1. Place the battery holder and LEDs on the fabric.
2. Arrange all LED positive (+) pins on one side and negative (−) pins on the other side.
3. Mark one continuous positive path from the battery (+) to each LED (+).
4. Mark one continuous negative path from the battery (−) to each LED (−).
5. Start at the battery’s positive (+) pin.
6. Sew to the first LED’s positive (+) pin and loop around it.
7. Continue sewing to the next LED positive pin without knotting off.
8. Repeat until the final LED positive pin, then knot off.
9. Use a separate piece of conductive thread for the negative path.
10. Sew from the battery’s negative (−) pin to each LED’s negative (−) pin.
11. Knot off, secure knots, and test the circuit.

## Checkpoint

All LEDs should light. If only one LED lights, check the connections to the other LEDs. If none light, check the battery, polarity, and whether the two thread paths are touching.

# Demonstration 3: Push Button Circuit

## Objective

Add a push button or switch so the LED circuit turns on only when the circuit is activated.

## Circuit idea

A push button acts like a gate in the circuit:

- not pressed = circuit open = LED off
- pressed = circuit closed = LED on

This makes the soft circuit interactive.

<div style="text-align: center; margin-bottom: 18px;">

## Steps

1. Place the battery holder, push button, and LED(s) on the fabric.
2. Mark the positive path from the battery (+) to one side of the push button.
3. Mark the next positive path from the other side of the button to the LED positive (+) pin.
4. Mark the negative path from the LED negative (−) pin back to the battery negative (−) pin.
5. Sew the first conductive path from battery (+) to the button.
6. Knot off neatly.
7. Use a new thread path from the other side of the button to the LED positive (+).
8. Sew the negative path from LED (−) to battery (−).
9. Secure all knots and trim loose thread.
10. Press the button to test the circuit.

## Checkpoint

The LED should turn on only when the button is pressed. If it is always on, check whether the button has been bypassed by touching thread. If it never turns on, check the button pins and thread contact.

## Troubleshooting guide

| Problem | Possible cause | What to check |
|---|---|---|
| LED does not light | Polarity is reversed | Check + and − connections |
| LED flickers | Weak contact | Loop conductive thread more tightly around pins |
| Circuit stops working | Positive and negative paths touch | Look for overlapping thread or loose ends |
| Only some LEDs light | Broken path or weak connection | Check each LED pin and stitch line |
| Button does not work | Wrong button pins used | Test with alligator clips first |
| Knot comes loose | Conductive thread is coarse | Secure knots with clear nail polish or adhesive |
| Fabric bends too much | Base fabric is unstable | Use felt, fusing, or embroidery hoop |

## Flex sensor

A **flex sensor** is a bend-sensitive sensor that changes its electrical behaviour when it is bent. In a soft circuit context, it can be useful for detecting movement in fabric-based or wearable projects, such as finger bending, wrist movement, posture changes, or fabric deformation.

In this tutorial, the flex sensor can be seen as a bridge between simple soft circuits and later sensor-based systems. While the basic LED circuit shows how electricity can create an output, a flex sensor introduces the idea of using a soft or flexible material as an input. In later tutorials, this kind of sensor can be connected to a Raspberry Pi or another microcontroller to read movement data, process it, and describe it using simple terms such as **straight**, **slightly bent**, **moderately bent**, or **strongly bent**.

## Summary

In this tutorial, we introduced soft circuits and e-textiles by building simple circuits with conductive thread, a power source, LEDs, and switches. The key idea is that electricity needs a closed path to flow, and the positive (+) and negative (−) thread lines must stay separate to avoid a short circuit.

This basic activity helps prepare us for later tutorials with electronic components, sensors, the Raspberry Pi, and EmotiBit. Understanding how power, connections, and simple interaction work in a soft circuit will make it easier to connect and test more advanced sensor-based systems later.
