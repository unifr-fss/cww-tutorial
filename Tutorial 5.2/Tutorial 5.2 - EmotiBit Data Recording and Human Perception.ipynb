## 2. EmotiBit Oscilloscope

The **EmotiBit Oscilloscope** is the main application used to connect to an EmotiBit over WiFi, view incoming signals in real time, and start or stop recording sessions.

When the EmotiBit blue LED is blinking, the device has connected to Wi-Fi. Open the EmotiBit Oscilloscope and select the device to confirm that live biometric data is streaming.

Open the **EmotiBit Oscilloscope** from the software bundle. The required network permissions and first connection were covered in Tutorial 5.1.

In simple terms, the Oscilloscope helps you do three main things:

1. **Connect** to the EmotiBit on the same WiFi network  
2. **Monitor** live biometric and motion signals on screen  
3. **Record** data for later analysis and experiments  

### What the main areas mean

<div style="text-align: center;">
  <img src="5.2 Pictures/emotibitoscilloscope.gif" alt="EmotiBitOscilloscopegif photo" width="900">
</div>

- **EmotiBit Device List**  
  This shows the EmotiBits detected on the network. Click the device name to connect. If no device is selected, the screen may show empty or flat channels. Once a device is selected, live data starts streaming.

- **Battery Level**  
  This shows the current battery percentage of the connected EmotiBit. It is better not to let the battery go too low during a session.

- **Record button**  
  This is used to start and stop a recording session.

- **SD File / recording status**  
  This area shows whether the EmotiBit is currently recording. If a recording session is active, the file name being recorded appears below the record section.

- **Clipping Events**  
  A clipping event means one or more sensor signals went outside expected bounds. This is a data-quality warning.

- **Overflow Events**  
  An overflow event means internal buffers overflowed and some samples may have been lost. Ideally, this should stay at or near zero.

- **Note / Log Note**  
  This area allows you to type labels or notes during a session. These notes are useful later when interpreting the recorded data.

- **Power Mode**  
  This controls EmotiBit power behavior. For normal beginner use, it is usually safest to remain in the default or normal mode unless you are specifically told otherwise.

- **Output List**  
  This controls optional output channels such as OSC, UDP, and LSL. For normal beginner recording, you usually do not need to change these settings.

### What the plotted channels mean

The exact channels visible can vary, but common examples include:

- **PPG:RED, PPG:IR, PPG:GRN** — optical pulse sensing channels  
- **HR** — heart rate  
- **EDA** — electrodermal activity / skin conductance  
- **ACC:X, ACC:Y, ACC:Z** — acceleration along three axes  
- **GYRO:X, GYRO:Y, GYRO:Z** — rotational motion along three axes  
- **MAG:X, MAG:Y, MAG:Z** — magnetometer signals  
- **TEMP1 / THERM** — temperature-related channels  
- **SCR:AMP, SCR:FREQ, SCR:RIS** — skin conductance response related values  

You do **not** need to fully interpret every channel at the beginning. For now, the important point is:

- if the EmotiBit is connected properly, you should see live values and changing traces  
- if nothing is connected, many channels may look flat, empty, or inactive  
- if the device is selected and worn properly, signal traces should update continuously in real time


## 3. Record a Session with the EmotiBit Oscilloscope

### How to start recording

1. Make sure the EmotiBit is powered on and connected to the same WiFi network as your computer.  
2. Open the **EmotiBit Oscilloscope**.  
3. In the **EmotiBit Device List**, click your EmotiBit to connect.  
4. Confirm that live signals are visible on the screen.  
5. Click the **Record** button on the top console.  
6. Once recording starts:
   - the record section becomes active
   - the EmotiBit red LED starts blinking
   - the SD file / recording section shows the file being recorded

### Good practice before and during recording

- wait a few seconds after connecting so that the stream becomes stable  
- check battery level  
- make sure the device is being worn correctly  
- add a short note during the recording using the **Note / Log Note** area; this creates a useful event marker that can later be seen alongside the recorded signals  
- watch that the live traces keep updating  
- large clipping or overflow counts may indicate data-quality issues  

### How to stop recording

To stop recording, click the **Record** button again.

After stopping:
- wait a few moments so the session is clearly finished  
- confirm that you are no longer recording  
- do not remove the battery or power off the board during an active recording  

### Carefully disconnecting after the session

1. **Stop the recording first** in the Oscilloscope.  
2. Wait a few moments so the session is clearly finished.  
3. If you are fully done, handle the board gently.  
4. If you remove power, do it carefully and avoid pulling on wires.  
5. If you remove the SD card, do it gently.

### General handling warnings

- do not force connectors  
- do not bend the stacked boards unnecessarily  
- do not remove the SD card roughly  
- do not unplug power in the middle of a firmware update or active flashing process  
- when wearing the EmotiBit, avoid unnecessary pulling, twisting, or pressure on the board and wires  

### How to confirm that data was recorded

After your session:

1. Make sure you had actually started and stopped recording properly in the Oscilloscope.  
2. Power down carefully if needed.  
3. Remove the SD card carefully.  
4. Insert the SD card into your computer.  
5. Open the SD card contents and check whether a new recording file is present.  
6. Compare file names or timestamps with your session time.

If a file was created during your session, that is your first confirmation that the recording was saved.

## 4. Human-Perception Questionnaire

Immediately after the recording, describe how the session felt **before viewing the processed sensor plots**.

The questionnaire focuses on four concepts that will later be compared with selected EmotiBit signals and fuzzy linguistic interpretations:

- heartbeat perception
- skin-temperature perception
- movement perception
- activation / arousal perception

Select one linguistic option for each question, then click **Save Questionnaire**.

> **Important:** There is no expected “correct” answer. The purpose is to preserve your own perception of the session so it can later be compared with sensor-based linguistic interpretations.


````python
from pathlib import Path
from datetime import datetime
import csv
import re
import ipywidgets as widgets
from IPython.display import display, clear_output

QUESTIONNAIRE_DIR = Path("..") / "tutorial_data" / "questionnaire"
QUESTIONNAIRE_DIR.mkdir(parents=True, exist_ok=True)

session_box = widgets.Text(placeholder="e.g. session_01", description="Session:", style={"description_width": "initial"})
heartbeat = widgets.RadioButtons(options=["much slower than usual","slightly slower than usual","normal / usual","slightly faster than usual","much faster than usual"])
skin_temperature = widgets.RadioButtons(options=["cool","slightly cool","comfortable / neutral","slightly warm","warm"])
movement = widgets.RadioButtons(options=["completely still","slightly moving","moderately moving","actively moving","very active"])
activation = widgets.RadioButtons(options=["very calm","calm","neutral / moderately alert","activated / alert","very activated / alert"])
save_button = widgets.Button(description="Save Questionnaire", button_style="success", icon="save")
output = widgets.Output()

def save_questionnaire(_):
    with output:
        clear_output()
        session_id = session_box.value.strip() or datetime.now().strftime("session_%Y%m%d_%H%M%S")
        safe_session_id = re.sub(r"[^A-Za-z0-9_-]+", "_", session_id).strip("_") or "session"
        response = {
            "session_id": session_id,
            "questionnaire_time": datetime.now().isoformat(timespec="seconds"),
            "heartbeat_perception": heartbeat.value,
            "skin_temperature_perception": skin_temperature.value,
            "movement_perception": movement.value,
            "activation_arousal_perception": activation.value,
        }
        output_path = QUESTIONNAIRE_DIR / f"questionnaire_{safe_session_id}.csv"
        with output_path.open("w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=response.keys())
            writer.writeheader()
            writer.writerow(response)
        print("Questionnaire saved successfully.")
        print(f"File: {output_path}")

save_button.on_click(save_questionnaire)

display(widgets.VBox([
    widgets.HTML("<h4>Recording Session</h4>"), session_box,
    widgets.HTML("<h4>How did your heartbeat feel during the recording?</h4>"), heartbeat,
    widgets.HTML("<h4>How did your skin/body temperature feel during the recording?</h4>"), skin_temperature,
    widgets.HTML("<h4>How much were you moving during the recording?</h4>"), movement,
    widgets.HTML("<h4>How activated or alert did you feel during the recording?</h4>"), activation,
    save_button, output
]))
````

## 5. Parse the Recorded SD-Card Data

After recording the data, you should see the original recorded files on the SD card, typically including a **CSV file** and a **JSON file**.

Before parsing, it is recommended to **copy these original recorded files from the SD card to a dedicated folder on your local computer**. This makes the files easier to manage and avoids working directly from the SD card.

### Why parsing is needed

The original recorded CSV file contains the full raw session data. The **EmotiBit Data Parser** processes this raw recorded file and generates **new CSV files for individual sensors and axes**, which can then be used for visualization and later analysis.

### Open the EmotiBit Data Parser

Inside the extracted **EmotiBit software bundle** for your operating system, locate and open the application named:

**EmotiBit Data Parser**

This application is separate from the **EmotiBit Oscilloscope**.

### Generate the parsed CSV files

1. Open **EmotiBit Data Parser**
2. In the top-left area of the window, click the box that says:

   **"Click here to load EmotiBit data file"**

3. Select the **original recorded CSV file**
4. The parser will then generate new parsed **CSV files** in the same folder

<div style="text-align: center;">
  <img src="5.2 Pictures/dataparser.png" alt="dataparser photo" width="750">
</div>

After parsing is completed, return to that folder and confirm that the new sensor-wise CSV files have been created successfully.

**Note:** After the new parsed CSV files are generated, it is recommended to move the original recorded CSV and JSON files to a separate folder, or clearly label them, so that the current working folder contains only the newly generated parsed CSV files. This helps avoid confusion during visualization, since the visualization step should use only the parsed CSV files, not the original recorded file.

At this point, the original recorded file has now been prepared for the next stage: **visualization**.


## 6. Visualize the Parsed Data

After parsing, the newly generated CSV files can be used for visualization.

For this step, it is recommended to use a **new virtual environment** and the **EmotiBit biometric library**.

Please follow only the subsection that matches your operating system.

<details>

  <br>

  On **macOS**, open a terminal and run the following commands:

  ~~~bash
  python3 -m venv emotibit_env
  source emotibit_env/bin/activate
  git clone https://github.com/EmotiBit/EmotiBit_Biometric_Lib.git
  cd EmotiBit_Biometric_Lib/py/

  pip install -e .
  ~~~

  ~~~bash
  python dataviewer_example.py \
    -p "/path/to/your/parsed/data/folder" \
    -n "base_filename_of_your_recording" \
    -u
  ~~~

  **What to replace:**
  - Replace `"/path/to/your/parsed/data/folder"` with the path to the folder where the newly generated parsed CSV files are located.
  - Replace `"base_filename_of_your_recording"` with the common recording name used in those parsed files.

  **Example:**  
  If the parsed CSV files are stored in a folder on your Desktop called `sensors`, and the files begin with:

  `2026-04-26_22-34-20-119626_...`

  then the final command would look like this:

  ~~~bash
  python dataviewer_example.py -p "/Users/your_username/Desktop/sensors" -n "2026-04-26_22-34-20-119626" -u
  ~~~

  Once this step is completed successfully, the visualization window should open and display the plotted sensor signals.

### 6.4 Understanding the Visualization

The figure below shows the visualization generated from the **parsed EmotiBit CSV files**. Each row represents one signal channel extracted from the recorded session.

<div style="text-align: center;">
  <img src="5.2 Pictures/visualization_plot.png" alt="visualization_plot photo" width="1200">
</div>

The labels in the plot are abbreviated channel names. In this visualization, they generally refer to the following:

- **EA** — electrodermal activity signal  
- **SF** — skin-response related derived signal  
- **PI** — PPG infrared channel  
- **PR** — PPG red channel  
- **PG** — PPG green channel  
- **HR** — heart rate  
- **TH** — thermopile / temperature-related signal  
- **T1** — temperature channel 1  
- **AX, AY, AZ** — accelerometer data on the x, y, and z axes  
- **GX, GY, GZ** — gyroscope data on the x, y, and z axes  
- **MX, MY, MZ** — magnetometer data on the x, y, and z axes  
- **DC EA** — event/tag related information shown together with the electrodermal activity view  

### What the highlighted red region means

The red highlighted region marks a **Log Note/ tagged or selected time interval** in the recording. It is not a separate sensor signal and it does not indicate an error. It is used to highlight a particular portion of the session across multiple channels at the same time.

### Important note about the plotted channels

Not every parsed CSV file is always shown as a separate visible plot. Some parsed files may contain timing, synchronization, helper information, or event-related data. This is normal.

The main purpose of this visualization is to help you inspect the recorded physiological and motion signals more clearly after parsing.
