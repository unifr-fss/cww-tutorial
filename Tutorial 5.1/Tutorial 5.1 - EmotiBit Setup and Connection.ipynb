# Tutorial 5.1: EmotiBit Setup and Connection

## Introduction

In this notebook, we begin working with the **EmotiBit** platform. The goal of this tutorial is to set up the EmotiBit hardware and software correctly and establish a successful live connection with the device.

## SAFETY WARNING

EmotiBit should NEVER be worn while electrically connected to any device that's attached to A/C wall power. Connecting EmotiBit USB or any other pin to another device can be a shock risk hazzard and should ONLY be done when EmotiBit is NOT in physical contact with the body.

This means:
- Do not wear the EmotiBit on the body while it is physically connected by USB or another electrical connection to a powered external device
- Only make wired electrical connections when the EmotiBit is **not** in physical contact with the body

This is an important safety rule and must always be followed.

## What is EmotiBit?

**EmotiBit** is a wearable biosensing platform that can be used to collect physiological and motion-related data. In this course, we use it as a sensing device that allows us to move from simple output-based interaction, such as LED control, toward real sensor-based data acquisition.

At this stage, the main purpose is not yet to interpret the recorded signals in depth. Instead, the aim is to learn how to prepare the device properly and make it operational in a reliable way.

## 1.1 Required Hardware

Before starting the setup, make sure the following components are available:

- EmotiBit board
- compatible Feather board
- lithium-ion battery
- microSD card
- microSD card reader
- USB cable
- computer running Windows, macOS, or Linux

Depending on the kit and the planned use case, additional items such as straps, electrodes, stickers, or barriers may also be included. However, for the basic setup and first data recording, the main focus is on the core hardware listed above.

## 1.3 General Handling Precautions

When handling the hardware, follow these precautions:

- do not force any connector into place
- check orientation carefully before stacking boards
- insert the battery connector gently but firmly
- handle the board and switches with care
- keep the components on a stable and clean surface during setup
- if something does not fit naturally, stop and recheck the alignment

The goal is to avoid physical damage while assembling the system.

# 2. Physical Setup and Configuration

In this section, we prepare the EmotiBit for first use.

The goal of this section is to:
- prepare the microSD card
- add Wi-Fi credentials
- understand important network limitations
- connect the battery
- check the hibernation switch
- stack the EmotiBit and Feather boards correctly

At the end of this section, the hardware should be physically prepared for software and firmware setup.


## 2.1 Preparation Before Assembly

Before physically assembling the EmotiBit, first make sure all the components mentioned earlier are ready:

It is best to prepare the microSD card and Wi-Fi configuration first, before completing the full hardware setup.

## 2.2 Prepare the microSD Card

Insert the microSD card into the card reader and connect it to your computer.

This microSD card will be used during the setup process, and it must contain the required configuration file before the EmotiBit is started normally.

At this stage, do not place the microSD card into the EmotiBit yet. First, prepare the configuration file on your computer.

## 2.3 Download the Wi-Fi Configuration File

Download the EmotiBit configuration file https://www.emotibit.com/files/config

**NOTE: For Wi-Fi connection, using a mobile hotspot is a convenient option, as it keeps the setup more convenient both at home and while travelling or working at the university.**

After downloading it:
- open the file in a plain text editor
- locate the Wi-Fi credentials section
- replace the placeholder network name with your Wi-Fi name
- replace the placeholder password with your Wi-Fi password

<div style="text-align: center;">
  <img src="5.1 Pictures/config-file-example.png" alt="config-file photo" width="500">
</div>

Make sure the file formatting remains correct while editing.
Do not remove quotation marks, brackets, or commas accidentally.

## 2.4 Save the Configuration File to the microSD Card

After editing the Wi-Fi credentials:

1. save the configuration file
2. copy or save it onto the microSD card
3. eject the microSD card safely from your computer

Only after this step is complete should the microSD card be inserted into the EmotiBit.

## 2.5 Wi-Fi Notes

Before continuing, note the following:

- EmotiBit currently supports **2.4 GHz Wi-Fi** 
- support for enterprise Wi-Fi is limited
- enterprise Wi-Fi support is only experimentally available for **ESP32 Feather** boards. 

For a first setup, it is best to use a simple and reliable 2.4 GHz Wi-Fi network whenever possible. which is also the band commonly used by many standard mobile hotspots, so **we suggest using the same mobile hotspot with same credentials that you used for the Raspberry Pi, as this makes the setup more convenient both at home and while travelling or working at the university**.

## 2.6 Insert the microSD Card into the EmotiBit

Now insert the prepared microSD card into the EmotiBit.

Before moving on, verify that:
- the microSD card is fully inserted
- the configuration file has already been saved to the card
- the card was safely ejected from the computer before use

This step is important because the EmotiBit uses the microSD card during startup and connection preparation.

## 2.7 Check the Hibernate Switch
Make sure the sliding switch (Hibernate switch) is set to the active (not HIB) position as shown.

- CAUTION: Excessive force can break the HIB switch. Handle with care! 

<div style="text-align: center;">
  <img src="5.1 Pictures/correctHibernateSwitch.jpg" alt="correctHibernateSwitch photo" width="300">
</div>


## 2.8 Connect the Battery

Now connect the lithium-ion battery to the Feather board.

While doing this:
- align the connector carefully
- insert it gently but firmly
- make sure the connector is pushed in fully
- do not force the connector if it does not align naturally

A loose battery connection can cause startup problems later.


## 2.9 Stack the EmotiBit and Feather Boards

Now stack the EmotiBit with the Feather board.

Check the alignment carefully before pushing the boards together.

The connector alignment must match correctly:
- the **12-pin connector** should go into the **12-pin socket**
- the **16-pin connector** should go into the **16-pin socket**

<div style="text-align: center;">
  <img src="5.1 Pictures/EmotiBit_stack_boards.gif" alt="EmotiBit stacking boards gif" width="300">
</div>

While stacking:
- align both sides carefully
- do not twist the boards
- do not force them together
- stop immediately if the fit does not look correct

This is one of the most important physical setup steps in the notebook.


## 2.10 Common Mistakes During Physical Setup

At this stage, the most common mistakes are:

- the Wi-Fi credentials were not edited correctly
- the configuration file was not saved to the microSD card
- the microSD card was not inserted properly
- the hibernation switch is still set to **HIB**
- the battery connector is loose
- the Feather and EmotiBit are not aligned correctly
- the wrong connector orientation was used during stacking

If something does not work later, return to this section and verify each item carefully.

  <br>

  Installing the Firmware

  The firmware is the software that runs on the EmotiBit hardware itself. You need to flash it once using a USB cable.

  Do NOT unplug the USB cable while flashing is in progress. This can permanently damage the Feather board.

  Connect the Feather to Your Laptop

  Connect the Feather ESP32 to your laptop using a data-capable USB cable.

  Check that the device is detected:

  `ls -la /dev/ttyUSB*`

  You should see /dev/ttyUSB0 in the output. If you do not see it, try a different USB cable (some cables are charge-only and cannot transfer data).

  Add Your User to the dialout Group

  This gives your user permission to communicate with USB serial devices:

  `sudo usermod -a -G dialout $USER`

  `newgrp dialout`

  Flash the Firmware

  Navigate to the firmware data folder:

  `cd ~/of_v0.11.2_linux64gcc6_release/addons/ofxEmotiBit/EmotiBitFirmwareInstaller/bin/data`

  Run the flash command:

  `./exec/linux/esptool --chip esp32 --port /dev/ttyUSB0 --baud 921600 \`

  `  --before default_reset --after hard_reset write_flash -z \`

  `  --flash_mode dio --flash_freq 80m --flash_size 4MB \`

  `  0x1000 ./esp32/EmotiBit_stock_firmware.ino.bootloader.bin \`

  `  0x8000 ./esp32/EmotiBit_stock_firmware.partitions.bin \`

  `  0xe000 ./esp32/boot_app0.bin \`

  `  0x10000 ./EmotiBit_stock_firmware.ino.feather_esp32.bin`

  A successful flash will end with: Hash of data verified. and Hard resetting via RTS pin...

  Disconnect the USB cable — the Feather no longer needs to be connected.

  <br>

  For Linux and Advanced Users

  Installing Emotibit Firmware on Feather ESP32 Huzzah

  Note: on macOS, you will need to give the EmotiBit FirmwareInatller application permissions to run the contained executables. To do so, right click on the EmotiBit FirmwareInstaller app and click on Open. This will open the application but you may close it to proceed below.

  The FirmwareInstaller essentaily uploads the latest EmotiBit FW onto the Feather

  We use the esptool command line tool to upload binary files to the feather.

  There are 2 requirements to run esptool

  - COM port on which the Feather is detected
  - The esptool bin file (provided in the software release).

  To perform the operations manually, the follow the below listed steps:

  - Navigate to the data folder located inside the EmotiBit software directory.
    - On Linux the path to the data folder should look like `EmotiBitSoftware-linux/ofxEmotiBit/EmotiBitFirmwareInstaller/bin/data`
    - On MacOS the path should look like `EmotiBitSoftware-macOS/EmotiBitFirmwareInstaller.app/Contents/Resources`
    - On Windows the path will be `C:\Program Files\EmotiBit\EmotiBit FirmwareInstaller\data`
  - Open a cmd prompt window for Windows or terminal for Linux/Mac at this location
  - Connect the Feather to the computer using a data-capable USB cable.
    - The Feather should show up as a COM port on the system.
    - On Windows: The device appears with a same similiar to `COM X` (where `X` is a number)
    - On mac/linux: You may find the COM port by running the terminal command `ls -la /dev/tty*`
    - Pro-tip for linux: the Feather may likely show up as `/dev/ttyUSB0`

  **WARNING: DO NOT UNPLUG OR RESET FEATHER WHILE UPLOAD/UPDATE IN PROGRESS. YOU COULD BRICK YOUR FEATHER!**

  replace `YOUR_FEATHER_PORT` with the COM port you detected in the previous step in the following command and run it to upload the firmware.

  `./exec/linux/esptool --chip esp32 --port YOUR_FEATHER_PORT --baud 921600 --before default_reset --after hard_reset write_flash -z --flash_mode dio --flash_freq 80m --flash_size 4MB 0x1000 ./esp32/EmotiBit_stock_firmware.ino.bootloader.bin 0x8000 ./esp32/EmotiBit_stock_firmware.partitions.bin 0xe000 ./esp32/boot_app0.bin 0x10000 ./EmotiBit_stock_firmware.ino.feather_esp32.bin`

  - [For linux] If you get a permission denied error, run the command `chmod u+x ./exec/esptool`, to make the file executable.
  - [For Windows] replace `./exec/linux/esptool` with `.\exec\win\esptool.exe`. You will also need to change all file paths to `.\esp32\name-of-file`
  - [For macOS] replace `./exec/linux/esptool` with `./exec/mac/esptool`.

## 4. Connect to EmotiBit and Verify the Live Stream

This section completes the setup by connecting the EmotiBit to the EmotiBit Oscilloscope and confirming that live sensor data is visible. Recording begins in Tutorial 6.

### 4.1 EmotiBit Oscilloscope Network Connection

The EmotiBit Oscilloscope communicates with the EmotiBit device over a WiFi network connection. For this communication to function, the Oscilloscope requires permission from your computer's operating system to access and use the local network.

### 4.2 Granting Network Permissions (Crucial Step)

The very first time you launch the EmotiBit Oscilloscope on your computer, your operating system may prompt you to grant network access permissions to the application.

**Action Required:** You must grant the appropriate network permissions when prompted.

⚠️ **Attention:** Without these permissions, the EmotiBit Oscilloscope cannot detect or communicate with the EmotiBit connected to your network. Data streaming will fail. Please follow the subsection that matches your operating system.

<details>

  - You may also allow access on **public networks** if you plan to use the EmotiBit on a public network.
  - If you have an antivirus program installed, make sure the appropriate settings are enabled to grant the EmotiBit Oscilloscope network access.
